## Creating a CNN architecture using pytorch

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class MNISTCNN(nn.Module):
    def __init__(self):
        super().__init__()
        
        self.conv1 = nn.Conv2d(
            in_channels = 1,
            out_channels = 8,
            kernel_size = 3
        )
        
        self.conv2 = nn.Conv2d(
            in_channels = 8,
            out_channels = 16,
            kernel_size = 3 
        )
        
        #the weight tensor hold (8,1,3,3)
        # 8 feature map, 1 channel, and kernel height * width = (3*3)
        
    def forward(self,x):
        x = self.conv1(x)
        x = F.relu(x)
        
        x = self.conv2(x)
        x = F.relu(x)
        
        return x

x = torch.randn(4,1,28,28)

model = MNISTCNN()
x = model(x)
print(model.conv1.weight.shape)
```

# CNN architecture for a RGB image

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

x = torch.randn(4, 3, 32, 32)

print(x.shape)
```

```python
class CIFAR10CNN(nn.Module):
    def __init__(self):
        super().__init__()
    
        # Layer 1
        self.conv1 = nn.Conv2d(
            in_channels = 3,
            out_channels = 16,
            kernel_size = 3,
            stride = 1,
            padding = 1  
        )
     
        self.pool = nn.MaxPool2d(
            kernel_size = 2,
            stride = 2 
        ) 
        #32 - 32-3+1 = 30*30 
        
        #Layer2
        self.conv2 = nn.Conv2d(
            in_channels = 16,
            out_channels = 32,
            kernel_size = 3,
            stride = 2
        ) 
        
        self.fc = nn.Linear(32*7*7,10)
         
    def forward(self,x):
        x = self.conv1(x)
        x = torch.relu(x)
        x = self.pool(x)
        
        x = self.conv2(x)
        x = torch.relu(x)
        
        x = torch.flatten(x,1)
        x = self.fc(x)
        
        return x
    
        
```

# Create the pipeline for the architecture

```python
model = CIFAR10CNN()
loss_fn = nn.CrossEntropyLoss() #For the classification task,uses log loss
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)
```

```python
x = torch.randn(4, 3, 32, 32)

#class
y = torch.tensor([3, 7, 1, 9])
```

```python
before = model.conv1.weight.detach().clone()
```

```python
optimizer.zero_grad()

# 2. Forward propagation
logits = model(x)

# 3. Calculate loss
loss = loss_fn(logits, y)

# 4. Backpropagation
loss.backward()

# 5. Gradient descent
optimizer.step()
```

```python
after = model.conv1.weight.detach().clone()

print(after[0] - before[0])
```
