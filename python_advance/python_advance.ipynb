```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
    
    def __add__(self,other):
        return Vector(self.x + other.x,self.y + other.y)
    
    def __sub__(self,other):
        return Vector(self.x - other.x,self.y - other.y)

v1 = Vector(1, 2)
v2 = Vector(3, 4)
v3 = Vector(2,1)
v4 = v1 + v2 + v3 
print(repr(v4))

v4 = v1 - v2 -v3
print(repr(v4))
```

# isinstance() function in python

```python
class Account:
    def __init__(self, balance):
        self.balance = balance

    def __eq__(self, other):
        return isinstance(other, Account) and self.balance == other.balance

    def __lt__(self, other):
        return self.balance < other.balance

a1 = Account(100)
a2 = Account(250)
print(a1 < a2)   # True (calls a1.__lt__(a2))
print(a1 == a2)  # False (calls a1.__eq__(a2))
```

# Container and Collection Protocols


```python
class CustomList:
    def __init__(self, *items):
        self._data = list(items)

    def __len__(self):
        return len(self._data)

    def __getitem__(self, index):
        return self._data[index]

cl = CustomList("apple", "banana", "cherry")
print(len(cl))      # 3
print(cl[1])        # banana
```

# Iteration and Generation

These methods make your object iterable in loops and generator contexts.

__iter__(self): Must return an iterator object (an object that implements __next__). Triggered by for x in obj or iter(obj).

__next__(self): Returns the next item in sequence or raises StopIteration.

```python
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        val = self.current
        self.current -= 1
        return val

for num in Countdown(3):
    print(num)  # Outputs 3, then 2, then 1
```

# 6. Context Managers (with Statement)

```python
class DatabaseConnection:
    def __enter__(self):
        print("Opening connection...")
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        print("Closing connection...")
        # Returning True suppresses any raised exception

with DatabaseConnection() as db:
    print("Querying data...")

# Output:
# Opening connection...
# Querying data...
# Closing connection...
```
